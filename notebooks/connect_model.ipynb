```python
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown
import os

load_dotenv()  # Load environment variables from .env file

api_key = os.getenv("OPENROUTER_API_KEY")

user_input =  "Làm cho tôi một bài thơ về tình yêu"

# The client automatically picks up the OPENAI_API_KEY environment variable
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key)

response = client.responses.create(
    model="deepseek/deepseek-v4-flash-0731:free", # or your preferred model like gpt-5.5 or gpt-4o
    input=user_input
)

res = response.output_text
```

```python
display(Markdown(f"**Response:** {res}"))
```

```python
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown
import os

load_dotenv()  # Load environment variables from .env file

api_key = os.getenv("OPENROUTER_API_KEY")


# The client automatically picks up the OPENAI_API_KEY environment variable
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key)

file_path = "../data/audio/test/file_am_thanh_1_kenh.wav"

with open(file_path, "rb") as audio_file:
    transcription = client.audio.transcriptions.create(
        model="meta/muse-voice-transcribe-1.0",
        file=audio_file,
        language="vi"  # Mã ngôn ngữ (ISO-639-1) giúp tăng độ chính xác và tốc độ
    )

print(transcription.text)
```

```python
display(Markdown(f"**Response:** {transcription.text}"))
```

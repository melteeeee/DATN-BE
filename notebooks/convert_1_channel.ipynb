```python
from pydub import AudioSegment

# 1. Chuyển đổi Stereo sang Mono
# Đọc file wav ban đầu
audio = AudioSegment.from_wav("../data/audio/test/TỔNG HỢP NỘI DUNG KHÓA TOÁN VẬT LÝ.wav")

# Chỉnh số lượng kênh về 1 (mono)
audio_mono = audio.set_channels(1).set_frame_rate(16000)

# Lưu lại thành file tạm hoặc ghi đè
mono_file_path = "../data/audio/test/file_am_thanh_1_kenh.wav"
audio_mono.export(mono_file_path, format="wav")
```
